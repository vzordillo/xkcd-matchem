# 07. Reaction Pathways & Nudged Elastic Band (NEB) Barriers

Whether you are summarizing a multi-step catalytic mechanism on a slide (`xm.plot_reaction_profile`) or visualizing a continuous minimum-energy path from an extended `.xyz` NEB trajectory (`xm.plot_neb`), `xkcd-matchem` turns relative energies into crisp, readable sketches.

## Basic Usage

Let's put an idealized two-step catalytic free-energy landscape next to the bundled 9-image $\text{Cl}^- + \text{CH}_3\text{Cl}$ $\text{S}_\text{N}2$ NEB trajectory (`sn2_neb.xyz`):

```python
import matplotlib.pyplot as plt

import xkcd_matchem as xm


neb_traj = xm.load_trajectory("sn2_neb.xyz")

with xm.xkcd_style("classic"):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.2, 4.4))
    xm.plot_reaction_profile(
        [0.0, 0.42, -0.35, 0.78, -0.85],
        labels=["R", "TS1", "Int", "TS2", "P"],
        ax=ax1,
        title="Catalytic Energy Profile (eV)",
    )
    xm.plot_neb(neb_traj, ax=ax2, title="SN2 NEB Barrier (eV)")
    fig.tight_layout()
```

## Advanced Usage

Presenting to organic or physical chemists who think in $\text{kcal/mol}$ instead of $\text{eV}$? Pass `unit="kcal/mol"` to convert energies automatically using CODATA 2022 constants, choose a custom curve color, and drop an xkcd-style annotation box onto the transition state with `xm.draw_callout()`:

```python
with xm.xkcd_style("classic"):
    fig, ax = plt.subplots(figsize=(6.8, 4.4))
    xm.plot_neb(
        neb_traj,
        ax=ax,
        unit="kcal/mol",
        color="#2A9D8F",
        title="SN2 NEB in kcal/mol with Callout",
    )
    xm.draw_callout(
        ax,
        "Walden inversion TS",
        xy=(0.50, 18.0),
        xytext=(0.15, 22.0),
        color="#D55E00",
    )
    fig.tight_layout()
```
