# 07. Reaction Pathways & Nudged Elastic Band (NEB) Barriers

Sketch step-wise catalytic free-energy diagrams (`plot_reaction_profile`) or continuous spline-interpolated NEB trajectories (`plot_neb`) from extended `.xyz` minimum-energy paths (`sn2_neb.xyz`).

## Basic Usage

Compare an idealized two-step catalytic profile alongside the real loaded $\text{S}_\text{N}2$ NEB barrier trajectory:

```python
import matplotlib.pyplot as plt
import xkcd_matchem as xm

neb_traj = xm.load_trajectory("sn2_neb.xyz")

with xm.xkcd_style("classic"):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.2, 4.4))
    xm.plot_reaction_profile(
        [0.0, 0.42, -0.35, 0.78, -0.85],
        labels=["R", "TS1", "Int", "TS2", "P"],
        ax=ax1,
        title="Catalytic Energy Profile (eV)",
    )
    xm.plot_neb(neb_traj, ax=ax2, title="SN2 NEB Barrier (eV)")
    fig.tight_layout()
```

## Advanced Usage

Convert NEB energies seamlessly to `kcal/mol` (`unit="kcal/mol"`), customize barrier colors, and overlay curved mechanistic arrows and callouts!

```python
with xm.xkcd_style("classic"):
    fig, ax = plt.subplots(figsize=(6.8, 4.4))
    xm.plot_neb(neb_traj, ax=ax, unit="kcal/mol", color="#2A9D8F", title="SN2 NEB in kcal/mol with Callout")
    xm.draw_callout(ax, "Walden inversion TS", xy=(0.50, 18.0), xytext=(0.15, 22.0), color="#D55E00")
    fig.tight_layout()
```
