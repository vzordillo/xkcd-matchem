# 9. Thermodynamics, EOS & ML Parity

Plot 2D binary convex hulls with above-hull decomposition lines, fit third-order Birch-Murnaghan equations of state, sketch Pourbaix diagrams, and report ML benchmark parity metrics.


```python
# Colab / isolated environment bootstrap (skipped if already installed)
try:
    import xkcd_matchem as xm
except ImportError:
    %pip install -q "xkcd-matchem @ git+https://github.com/vzordillo/xkcd-matchem.git"
    import xkcd_matchem as xm

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig, axes = plt.subplots(1, 3, figsize=(13.2, 4.2))
xm.plot_phase_diagram_2d([0.0, 0.25, 0.5, 0.75, 1.0], [0.0, -0.35, -0.46, -0.05, 0.0], ["A", "A3B", "AB", "AB3", "B"], ax=axes[0], title="Binary Convex Hull")
vols = np.linspace(36.0, 44.0, 8)
energies = -12.4 + 0.065 * (vols - 40.0) ** 2
xm.plot_eos(vols, energies, ax=axes[1], title="Birch-Murnaghan EOS")
rng = np.random.default_rng(7)
yt = np.linspace(-3.0, 1.0, 24)
yp = yt + rng.normal(0.0, 0.1, size=24)
xm.plot_parity(yt, yp, ax=axes[2], title="ML Energy Parity")
plt.tight_layout()
```
