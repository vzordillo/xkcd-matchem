# 09. Thermodynamics, Phase Stability & ML Parity Plots

Materials discovery talks often pull together three complementary summaries: thermodynamic stability (convex hulls and Pourbaix maps), elastic response (equations of state), and model accuracy (ML interatomic potential parity plots or functional benchmarks).

Let's see how cleanly they sit together in one cohesive xkcd-style layout.

## Basic Usage

In three lines, compute and sketch a binary convex hull (`plot_phase_diagram_2d`), fit equilibrium volume $V_0$ and bulk modulus $B_0$ via a third-order Birch-Murnaghan Equation of State (`plot_eos`), and display MAE / RMSE / $R^2$ parity metrics (`plot_parity`):

```python
import matplotlib.pyplot as plt
import numpy as np
import xkcd_matchem as xm

vols = np.linspace(36.0, 44.0, 8)
energies = -12.4 + 0.065 * (vols - 40.0) ** 2
rng = np.random.default_rng(12)
y_true = np.linspace(-3.0, 1.0, 24)
y_pred = y_true + rng.normal(0.0, 0.11, size=24)

with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(1, 3, figsize=(15.0, 4.4))
    xm.plot_phase_diagram_2d(
        [0.0, 0.25, 0.5, 0.75, 1.0],
        [0.0, -0.34, -0.45, -0.05, 0.0],
        ["A", "A3B", "AB", "AB3", "B"],
        ax=axes[0],
        title="Binary Convex Hull",
    )
    xm.plot_eos(vols, energies, ax=axes[1], title="Birch-Murnaghan EOS")
    xm.plot_parity(y_true, y_pred, ax=axes[2], title="MLIP Energy Parity")
    fig.tight_layout()
```

## Advanced Usage

Combine electrochemical Pourbaix stability regions (`plot_pourbaix`) with categorical functional benchmarks (`plot_bar_comparison`), choosing built-in palettes like `"pastel"` and `"vibrant"` or passing your own color list:

```python
with xm.xkcd_style("classic", palette=["#264653", "#2A9D8F", "#E9C46A", "#F4A261"]):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.2, 4.6))
    xm.plot_pourbaix(
        [
            ("M2+", [(0, -0.5), (4.5, -0.5), (4.5, 1.5), (0, 1.5)]),
            ("MO2", [(4.5, -0.5), (14, -0.5), (14, 1.5), (4.5, 1.5)]),
        ],
        ax=ax1,
        palette="pastel",
        title="Custom Pourbaix Regions",
    )
    xm.plot_bar_comparison(
        ["PBE", "PBE+U", "r2SCAN", "HSE06"],
        [0.62, 1.48, 1.04, 1.15],
        ylabel="Band Gap (eV)",
        palette="vibrant",
        ax=ax2,
        title="Functional Band Gap Comparison",
    )
    fig.tight_layout()
```
