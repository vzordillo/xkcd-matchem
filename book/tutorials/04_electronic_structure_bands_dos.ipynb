# 4. Electronic Structure: Band Structures & DOS

Parse Quantum ESPRESSO (`bands.dat`, `dos.dat`) and VASP (`EIGENVAL`, `DOSCAR`) outputs directly and combine them in synchronized panels.


```python
# Colab / isolated environment bootstrap (skipped if already installed)
try:
    import xkcd_matchem as xm
except ImportError:
    %pip install -q "xkcd-matchem @ git+https://github.com/vzordillo/xkcd-matchem.git"
    import xkcd_matchem as xm

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

```python
bands = xm.load("si_qe_bands.dat")
dos = xm.load("si_qe_dos.dat")
bands.fermi_energy = dos.fermi_energy
bands.k_ticks = [0.0, 0.4, 0.8, 1.2]
bands.k_labels = ["L", "Gamma", "X", "K"]
fig, _ = xm.plot_bands_dos(bands, dos, ylim=(-8.0, 9.0), title="Silicon Band Dispersion & DOS (QE)")
```

## Orbital-Projected & Spin-Polarized DOS


```python
e = np.linspace(-6.0, 4.0, 200)
up = 1.5 * np.exp(-((e + 1.8) / 0.7) ** 2) + 0.9 * np.exp(-((e - 1.4) / 0.6) ** 2)
dn = 1.3 * np.exp(-((e + 1.2) / 0.7) ** 2) + 1.1 * np.exp(-((e - 1.8) / 0.6) ** 2)
sp_dos = xm.DensityOfStates(
    energies=e,
    total_dos=up,
    spin_down_dos=dn,
    projected_dos={"Fe 3d": 0.75 * up},
)
xm.plot_dos(sp_dos, title="Spin-Polarized & Projected DOS");
```
