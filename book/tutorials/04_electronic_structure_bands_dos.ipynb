# 04. Electronic Band Structures & Density of States

`xkcd-matchem` parses Quantum ESPRESSO (`bands.dat`, `dos.dat`, `pw.out`) and VASP (`EIGENVAL`, `DOSCAR`) natively without external parsers. `BandStructure.band_gap()` accounts for spin channels, optional occupation weights, and continuous Fermi-level crossings (returning `0.0 eV` cleanly for metals!).

## Basic Usage

Plot individual band dispersion, density of states, or a Fermi-aligned side-by-side `plot_bands_dos(...)` composite panel in one line:

```python
import xkcd_matchem as xm

bands = xm.load("si_qe_bands.dat")
dos = xm.load("si_qe_dos.dat")
print(f"Estimated band gap from sampled eigenvalues: {bands.band_gap():.2f} eV")

fig, (ax_b, ax_d) = xm.plot_bands_dos(
    bands,
    dos,
    ylim=(-8.0, 10.0),
    title="Silicon Band Structure & DOS (QE)",
)
```

## Advanced Usage

Customize the theme (`dark_chalk`), override series colors (`palette=...`), and verify metallic crossing detection (`BandStructure.band_gap()` returns `0.0` when a single band crosses $E_F$ even if sampled discrete points avoid zero):

```python
import numpy as np

# Verify continuous Fermi-crossing gap detection on a synthetic metal band
metal = xm.BandStructure(
    k_distances=np.array([0.0, 0.5, 1.0]),
    eigenvalues=np.array([[-1.2, 0.8], [-0.4, 1.2], [0.6, 1.6]])[None, :, :],
    fermi_energy=0.0,
)
print("Metal continuous-crossing gap (expected 0.0):", metal.band_gap())

fig, _ = xm.plot_bands_dos(
    bands,
    dos,
    ylim=(-6.0, 8.0),
    palette=["#56B4E9", "#F0E442", "#009E73"],
    preset="dark_chalk",
    title="Silicon Bands + DOS ('dark_chalk' Slide Theme)",
)
```
