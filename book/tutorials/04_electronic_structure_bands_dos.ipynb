# 04. Electronic Band Structures & Density of States

Two plots tell most of a crystal's electronic story:
- An **electronic band structure** traces the allowed energies of electrons as their crystal momentum (**k**-vector) moves along high-symmetry directions inside the crystal's Brillouin zone. The **Fermi level $E_F$** marks the boundary between occupied and empty states at 0 K: if bands cross $E_F$, the material conducts like a metal; if an energy range around $E_F$ has no bands at all, that empty window is the **band gap** of a semiconductor or insulator.
- The **density of states (DOS)** counts how many electronic states sit at each energy across the whole crystal, making flat high-density bands and band edges jump out visually.

`xkcd-matchem` reads Quantum ESPRESSO (`bands.dat`, `dos.dat`, `pw.out`) and VASP (`EIGENVAL`, `DOSCAR`, `PROCAR`) outputs natively with zero external dependencies, and aligns their energy axes automatically around $E_F$. Behind the scenes, `BandStructure.band_gap()` checks spin channels, occupation weights (when present), and continuous band crossings—so a metallic band that crosses $E_F$ between two sampled $k$-points still reports `0.0 eV` cleanly instead of a false nonzero gap.

## Basic Usage

Load silicon's band path (`si_qe_bands.dat`) and density of states (`si_qe_dos.dat`), check the estimated gap, and sketch a linked side-by-side `plot_bands_dos()` figure in one call:

```python
import xkcd_matchem as xm


bands = xm.load("si_qe_bands.dat")
dos = xm.load("si_qe_dos.dat")
print(f"Estimated band gap from sampled eigenvalues: {bands.band_gap():.2f} eV")

fig, (ax_b, ax_d) = xm.plot_bands_dos(
    bands,
    dos,
    ylim=(-8.0, 10.0),
    title="Silicon Band Structure & DOS (QE)",
)
```

## Advanced Usage

Need a dark chalkboard slide for a seminar talk? Switch `preset="dark_chalk"` and pass a high-contrast `palette=[...]`. Below we also construct a synthetic 2-band metal to verify that `band_gap()` catches continuous $E_F$ crossings between discrete $k$-points:

```python
import numpy as np


# Verify continuous Fermi-crossing gap detection on a synthetic metal band
metal = xm.BandStructure(
    k_distances=np.array([0.0, 0.5, 1.0]),
    eigenvalues=np.array([[-1.2, 0.8], [-0.4, 1.2], [0.6, 1.6]])[None, :, :],
    fermi_energy=0.0,
)
print("Metal continuous-crossing gap (expected 0.0):", metal.band_gap())

fig, _ = xm.plot_bands_dos(
    bands,
    dos,
    ylim=(-6.0, 8.0),
    palette=["#56B4E9", "#F0E442", "#009E73"],
    preset="dark_chalk",
    title="Silicon Bands + DOS ('dark_chalk' Slide Theme)",
)
```
