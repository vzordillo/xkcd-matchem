# 2. Molecules in 3D & 2D

Render organic molecules from `.xyz`, `.sdf`, `.pdb`, `.gjf`, or `.out` files in depth-sorted 3D or PCA-aligned 2D skeletal representations.


```python
# Colab / isolated environment bootstrap (skipped if already installed)
try:
    import xkcd_matchem as xm
except ImportError:
    %pip install -q "xkcd-matchem @ git+https://github.com/vzordillo/xkcd-matchem.git"
    import xkcd_matchem as xm

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

## Comparing 3D Render Styles & Color Schemes


```python
aspirin = xm.load("aspirin.xyz")
fig, axes = plt.subplots(1, 3, figsize=(12.5, 4.0))
xm.plot_molecule(aspirin, ax=axes[0], style="ball_and_stick", palette="xkcd", show_legend=False, title="Ball & Stick (xkcd)")
xm.plot_molecule(aspirin, ax=axes[1], style="spacefill", palette="jmol", show_legend=False, title="vdW Spacefill (Jmol)")
xm.plot_molecule(aspirin, ax=axes[2], style="stick", palette="colorblind", show_legend=False, title="Stick (Colorblind)")
plt.tight_layout()
```

## 2D PCA Skeletal Projection & Bond Distance/Angle Callouts


```python
benzene = xm.load("benzene.sdf")
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10.5, 4.2))
xm.plot_molecule(benzene, ax=ax1, mode="2d", title="Benzene 2D Skeletal (V2000 SDF)")
xm.plot_molecule(aspirin, ax=ax2, mode="3d", show_legend=False, title="Distance Annotation")
xm.annotate_distance(ax2, aspirin, 0, 1)
xm.annotate_angle(ax2, aspirin, 0, 1, 2)
plt.tight_layout()
```
