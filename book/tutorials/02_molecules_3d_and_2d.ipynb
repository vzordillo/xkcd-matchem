# 02. Molecules in 3D, 2D & Interactive Browser Studio

Molecular plots in `xkcd-matchem` combine depth-sorted split half-bonds (colored by each bonded atom), crisp white specular dots, and clean borderless inline legends—so molecules stay legible even when you drop them onto a crowded slide or poster.

In this notebook, you will work with three bundled structures—**aspirin** (`aspirin.xyz`), **benzene** (`benzene.sdf`), and **alanine dipeptide** (`alaninedipeptide.pdb`)—and launch the offline Interactive 3D Studio right inside the notebook.

## Basic Usage

`xm.plot_molecule()` supports three 3D styles (`"ball_and_stick"`, `"spacefill"`, `"stick"`) plus a flat 2D skeletal mode (`mode="2d"`). When a file like `benzene.sdf` stores explicit bond orders, double and aromatic rings are preserved automatically; when a file like `aspirin.xyz` has only coordinates, bonds are inferred cleanly from covalent radii.

```python
import matplotlib.pyplot as plt
import xkcd_matchem as xm

aspirin = xm.load("aspirin.xyz")
benzene = xm.load("benzene.sdf")

with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(1, 3, figsize=(14.0, 4.5))
    xm.plot_molecule(aspirin, ax=axes[0], style="ball_and_stick", title="Aspirin 3D (Ball & Stick)")
    xm.plot_molecule(aspirin, ax=axes[1], style="spacefill", title="Aspirin 3D (Spacefill)")
    xm.plot_molecule(benzene, ax=axes[2], mode="2d", title="Benzene 2D Skeletal")
    fig.tight_layout()
```

## Advanced Usage

For fine control over framing and annotations, drop down to `xm.render_structure_3d()`:
- Set exact Euler angles (`azimuth`, `elevation`, `roll`) and `perspective` depth on `xm.Camera`.
- Pass an inline `{symbol: "#HEX"}` dictionary directly to `palette=` to highlight functional groups.
- Measure and annotate real geometric bond lengths (`xm.annotate_distance`) and bond angles (`xm.annotate_angle`) directly on the projected 3D scene.

```python
pep = xm.load("alaninedipeptide.pdb")
cam = xm.Camera(azimuth=-108.0, elevation=28.0, roll=8.0, perspective=0.55)
custom_colors = {"C": "#4A5568", "N": "#2B6CB0", "O": "#E53E3E", "H": "#F7FAFC"}

with xm.xkcd_style("classic"):
    fig, ax = plt.subplots(figsize=(6.4, 5.2))
    xm.render_structure_3d(
        pep,
        ax=ax,
        camera=cam,
        palette=custom_colors,
        show_cell=False,
        show_labels=False,
        radius_scale=1.05,
    )
    xm.render.draw_element_legend(ax, pep.symbols, scheme=custom_colors)
    d_val = xm.annotate_distance(ax, pep, 0, 1, camera=cam, color="#D55E00")
    ang_val = xm.annotate_angle(ax, pep, 0, 1, 2, camera=cam, color="#009E73")
    ax.set_title(f"Alanine Dipeptide | d(0-1)={d_val:.2f} A, angle={ang_val:.1f} deg")
    fig.tight_layout()
```
