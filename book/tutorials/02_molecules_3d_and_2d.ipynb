# 02. Molecules in 3D, 2D & Interactive Browser Studio

Molecular renderers in `xkcd-matchem` draw depth-sorted split half-bonds directly behind endpoint atom spheres, pure white circular specular highlights, and clean borderless inline legends.

In this notebook we explore aspirin (`aspirin.xyz`), benzene (`benzene.sdf`), and alanine dipeptide (`alaninedipeptide.pdb`), plus launch an offline interactive 3D studio right inside the page!

## Basic Usage

Switch effortlessly between `"ball_and_stick"`, `"spacefill"`, `"stick"`, and 2D skeletal projections (`mode="2d"`).

```python
import matplotlib.pyplot as plt
import xkcd_matchem as xm

aspirin = xm.load("aspirin.xyz")
benzene = xm.load("benzene.sdf")

with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(1, 3, figsize=(14.0, 4.5))
    xm.plot_molecule(aspirin, ax=axes[0], style="ball_and_stick", title="Aspirin 3D (Ball & Stick)")
    xm.plot_molecule(aspirin, ax=axes[1], style="spacefill", title="Aspirin 3D (Spacefill)")
    xm.plot_molecule(benzene, ax=axes[2], mode="2d", title="Benzene 2D Skeletal")
    fig.tight_layout()
```

### Interactive 3D Studio Right in Jupyter

Call `xm.view_structure(...)` with the default `"auto"` or `"html"` backend to spin, zoom, re-color individual chemical elements live, download PNG/SVG, and copy reproducible Python snippets:

```python
xm.view_structure(
    "aspirin.xyz",
    style="ball_and_stick",
    camera=xm.Camera(azimuth=-112.0, elevation=22.0, perspective=0.35),
)
```

## Advanced Usage

Customize camera perspective/roll, highlight functional groups using inline `{symbol: color}` overrides, and overlay sketchy bond length (`annotate_distance`) or bond angle (`annotate_angle`) callouts.

```python
pep = xm.load("alaninedipeptide.pdb")
cam = xm.Camera(azimuth=-108.0, elevation=28.0, roll=8.0, perspective=0.55)
custom_colors = {"C": "#4A5568", "N": "#2B6CB0", "O": "#E53E3E", "H": "#F7FAFC"}

with xm.xkcd_style("classic"):
    fig, ax = plt.subplots(figsize=(6.4, 5.2))
    xm.render_structure_3d(
        pep,
        ax=ax,
        camera=cam,
        palette=custom_colors,
        show_cell=False,
        show_labels=False,
        radius_scale=1.05,
    )
    xm.render.draw_element_legend(ax, pep.symbols, scheme=custom_colors)
    d_val = xm.annotate_distance(ax, pep, 0, 1, camera=cam, color="#D55E00")
    ang_val = xm.annotate_angle(ax, pep, 0, 1, 2, camera=cam, color="#009E73")
    ax.set_title(f"Alanine Dipeptide | d(0-1)={d_val:.2f} A, angle={ang_val:.1f} deg")
    fig.tight_layout()
```
