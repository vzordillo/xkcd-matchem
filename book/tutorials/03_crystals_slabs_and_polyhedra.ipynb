# 03. Crystals, Slabs, Miller Planes & Polyhedra

`xkcd-matchem` renders periodic crystals using depth-sorted back dashed edges, a subtle translucent base plane, solid front cell edges, and clean exterior `a, b, c` lattice labels.

For surface slabs (`plot_slab`), it automatically detects large vacuum gaps along $c$:
- `view="side"` draws an orthographic `(b, c)` slab profile clipped cleanly inside shaded slab & vacuum bands.
- `view="isometric"` trims excess vacuum around the slab while calling out the true DFT vacuum thickness.

## Basic Usage

Let's visualize diamond Si with a `(111)` Miller slice, a `2x2x1` $\text{2H-MoS}_2$ monolayer supercell, a `TiO2 (110)` slab side profile, and $\text{TiO}_6$ coordination polyhedra in rutile `TiO2`!

```python
import matplotlib.pyplot as plt
import xkcd_matchem as xm

with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(2, 2, figsize=(12.2, 9.6))
    xm.plot_crystal("si_diamond.cif", miller_planes=[(1, 1, 1)], ax=axes[0, 0], title="Diamond Si + (111)")
    xm.plot_crystal("mos2_2h.poscar", supercell=(2, 2, 1), ax=axes[0, 1], title="2H-MoS2 (2x2x1)")
    xm.plot_slab("tio2_slab.poscar", supercell=(3, 2, 1), view="side", ax=axes[1, 0], title="TiO2 (110) Side Profile")
    xm.plot_coordination("tio2_slab.poscar", "Ti", "O", supercell=(2, 2, 1), ax=axes[1, 1], title="TiO2 (110) Polyhedra")
    fig.tight_layout()
```

### Interactive Crystal & Polyhedra Exploration

Orbit around the $\text{TiO}_2$ coordination polyhedra or switch between orthographic and perspective projections directly in the embedded viewer:

```python
xm.view_structure(
    "tio2_slab.poscar",
    style="polyhedra",
    polyhedra=("Ti", "O"),
    supercell=(2, 2, 1),
    camera="perspective",
)
```

## Advanced Usage

Combine `Camera.from_preset(...)` (`"isometric"`, `"perspective"`, `"top"`, `"a"`, `"b"`, `"c"`, `"c*"`), explicit `center_in_cell()`, `show_cell="wireframe"` / `"clean"`, and custom element palettes!

```python
mos2 = xm.load("mos2_2h.poscar").center_in_cell(axes=(0, 1, 2))
top_cam = xm.Camera.from_preset("c*", cell=mos2.cell, zoom=1.08)
persp_cam = xm.Camera.from_preset("perspective", zoom=1.05)

with xm.xkcd_style("classic"):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.0, 5.0))
    xm.plot_crystal(
        mos2,
        supercell=(2, 2, 1),
        camera=top_cam,
        show_cell="clean",
        palette={"Mo": "#2B6CB0", "S": "#ECC94B"},
        ax=ax1,
        title="Along Reciprocal c* (Honeycomb Layer)",
    )
    xm.plot_slab(
        "tio2_slab.poscar",
        supercell=(2, 2, 1),
        view="isometric",
        vacuum_margin=2.2,
        camera=persp_cam,
        ax=ax2,
        title="TiO2 Slab: Trimmed Isometric View",
    )
    fig.tight_layout()
```
