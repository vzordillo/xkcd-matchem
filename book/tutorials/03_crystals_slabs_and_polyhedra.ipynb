# 3. Crystals, Slabs & Coordination Polyhedra

Load CIFs (with space-group symmetry expansion) and VASP POSCARs, repeat supercells, highlight `(hkl)` Miller slices, and extract `ConvexHull` coordination polyhedra.


```python
# Colab / isolated environment bootstrap (skipped if already installed)
try:
    import xkcd_matchem as xm
except ImportError:
    %pip install -q "xkcd-matchem @ git+https://github.com/vzordillo/xkcd-matchem.git"
    import xkcd_matchem as xm

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

## Diamond Si `(111)` Plane & 2H-MoS2 Supercell


```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.5, 4.8))
xm.plot_crystal("si_diamond.cif", miller_planes=[(1, 1, 1)], ax=ax1, title="Si Diamond + (111) Plane")
xm.plot_crystal("mos2_2h.poscar", supercell=(2, 2, 1), ax=ax2, title="2H-MoS2 2x2x1 Supercell")
plt.tight_layout()
```

## Rutile TiO2 (110) Slab & `TiO6` Coordination Polyhedra


```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.5, 5.0))
xm.plot_slab("tio2_slab.poscar", ax=ax1, title="TiO2 (110) Surface Slab")
xm.plot_coordination("tio2_slab.poscar", center="Ti", ligand="O", ax=ax2, title="Ti-O Coordination Polyhedra")
plt.tight_layout()
```
