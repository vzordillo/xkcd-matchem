# 03. Crystals, Slabs, Miller Planes & Polyhedra

Periodic materials introduce a few extra visual puzzles—the repeating **unit cell** box can obscure atoms, sites sitting on periodic faces can look split apart, and density functional theory (DFT) **surface slabs** (thin crystal slices separated by empty space to model an exposed surface) often include 10–15 Å of vacuum that squishes the interesting surface atoms.

`xkcd-matchem` handles these cleanly out of the box:
- **Unit cells** draw dashed back edges, a subtle translucent floor plane, solid front edges, and exterior `a, b, c` lattice-vector labels.
- **Miller planes `(hkl)`** slice cleanly through the unit cell at the exact crystallographic plane intercepts (for example, `(111)` cuts diagonally through all three axes of cubic silicon).
- **Coordination polyhedra** (`xm.plot_coordination`) wrap each central metal atom's neighboring ligands in a translucent geometric cage—like the $\text{TiO}_6$ octahedra below—making local bonding geometry easy to read at a glance.
- **Surface slabs** (`xm.plot_slab`) automatically locate the vacuum gap along $c$—so `view="side"` frames the slab and vacuum in crisp shaded bands, while `view="isometric"` trims dead white space and annotates the real DFT vacuum thickness.

## Basic Usage

Here are four everyday materials-science views in four single function calls: a diamond Si `(111)` Miller slice with a standalone corner crystal-axes triad (`crystal_axes="corner"`), a `2x2x1` $\text{2H-MoS}_2$ monolayer supercell, a $\text{TiO}_2(110)$ slab side profile, and bond-free $\text{TiO}_6$ coordination polyhedra:

```python
import matplotlib.pyplot as plt

import xkcd_matchem as xm


with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(2, 2, figsize=(12.2, 9.6))
    xm.plot_crystal(
        "si_diamond.cif",
        miller_planes=[(1, 1, 1)],
        crystal_axes="corner",
        ax=axes[0, 0],
        title="Diamond Si + (111)",
    )
    xm.plot_crystal(
        "mos2_2h.poscar",
        supercell=(2, 2, 1),
        ax=axes[0, 1],
        title="2H-MoS2 (2x2x1)",
    )
    xm.plot_slab(
        "tio2_slab.poscar",
        supercell=(3, 2, 1),
        view="side",
        ax=axes[1, 0],
        title="TiO2 (110) Side Profile",
    )
    xm.plot_coordination(
        "tio2_slab.poscar",
        "Ti",
        "O",
        supercell=(2, 2, 1),
        ax=axes[1, 1],
        title="TiO2 (110) Polyhedra",
    )
    fig.tight_layout()
```

### Explore Coordination Polyhedra Interactively

Pass `style="polyhedra"` and `polyhedra=("Ti", "O")` to `xm.view_structure()` to orbit around the shaded octahedra and switch between `Iso`, `Persp`, and crystallographic `a, b, c, a*, b*, c*` presets right in your browser:

```python
xm.view_structure(
    "tio2_slab.poscar",
    style="polyhedra",
    polyhedra=("Ti", "O"),
    supercell=(2, 2, 1),
    camera="perspective",
)
```

## Advanced Usage

Two helpers make layered and surface structures look noticeably sharper:
- `.center_in_cell(axes=(0, 1, 2))` shifts fractional coordinates away from periodic boundaries so layers sit cleanly inside the box.
- `xm.Camera.from_preset("c*", cell=mos2.cell)` aligns the camera strictly down the reciprocal $c^*$ axis—essential for looking straight onto the basal plane of hexagonal or monoclinic cells where $c$ is not perpendicular to the $(a, b)$ layer!

```python
mos2 = xm.load("mos2_2h.poscar").center_in_cell(axes=(0, 1, 2))
top_cam = xm.Camera.from_preset("c*", cell=mos2.cell, zoom=1.08)
persp_cam = xm.Camera.from_preset("perspective", zoom=1.05)

with xm.xkcd_style("classic"):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.0, 5.0))
    xm.plot_crystal(
        mos2,
        supercell=(2, 2, 1),
        camera=top_cam,
        show_cell=False,
        crystal_axes="corner",
        palette={"Mo": "#2B6CB0", "S": "#ECC94B"},
        ax=ax1,
        title="Along Reciprocal c* (Separate Crystal Axes)",
    )
    xm.plot_slab(
        "tio2_slab.poscar",
        supercell=(2, 2, 1),
        view="isometric",
        vacuum_margin=2.2,
        camera=persp_cam,
        ax=ax2,
        title="TiO2 Slab: Trimmed Isometric View",
    )
    fig.tight_layout()
```
