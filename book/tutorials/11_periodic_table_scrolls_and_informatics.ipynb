## Basic Usage

### 1. The Periodic Table of Elements (Families, Heatmaps & Spotlights)

Plot all 118 elements by chemical family, map any built-in property (`electronegativity`, `covalent_radius`, `vdw_radius`, `atomic_mass`) or custom `{symbol: value}` dict with hatched missing tiles (`missing_shade="hatch"`), or spotlight the elements of a compound like `"LiFePO4"`:

```python
import matplotlib.pyplot as plt
import xkcd_matchem as xm

with xm.xkcd_style("classic"):
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(13.5, 12.2))
    xm.plot_periodic_table(
        element_property="electronegativity",
        cmap="YlOrRd",
        missing_shade="hatch",
        ax=ax1,
        title="Pauling Electronegativity Heatmap (Hatched Missing Elements)",
    )
    xm.plot_periodic_table(
        highlight="LiFePO4",
        missing_shade="dots",
        ax=ax2,
        title="Formula Spotlight: LiFePO4 Cathode Elements",
    )
    fig.tight_layout(h_pad=1.5)
```

### 2. Background Grid Papers & Hatched Shading

Pass `background_grid=` (`"ruled"`, `"graph"`, `"dotgrid"`, `"isometric"`, `"blueprint"`) and `shade=` (`"hatch"`, `"crosshatch"`, `"dots"`, `"cycle"`, or raw Matplotlib hatch patterns) directly to plotters—or apply them to any Matplotlib `Axes` via `xm.apply_background_grid()`:

```python
with xm.xkcd_style("classic"):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.6, 4.4))
    xm.plot_bar_comparison(
        ["PBE", "r2SCAN", "HSE06", "GW"],
        [0.62, 1.04, 1.15, 1.24],
        ylabel="Si Band Gap (eV)",
        shade="cycle",
        background_grid="graph",
        ax=ax1,
        title="Hatched Bars on Graph Paper",
    )
    xm.plot_pourbaix(
        [
            ("Ti3+", [(0, -0.8), (3.8, -0.8), (3.8, 1.4), (0, 1.4)]),
            ("TiO2", [(3.8, -0.8), (14, -0.8), (14, 1.4), (3.8, 1.4)]),
        ],
        shade=("crosshatch", "dots"),
        background_grid="dotgrid",
        ax=ax2,
        title="Crosshatch & Stipple Pourbaix on Dot Grid",
    )
    fig.tight_layout()
```

## Advanced Usage

### 3. Old Parchment Scroll Cards (Portrait & Landscape)

Use `xm.plot_scroll_card(...)` standalone or `board.add_scroll_card(...)` inside a poster `Board` to turn checklists, lab protocols, synthesis roadmaps, or maps into vintage scrolled parchment cards:

```python
board = xm.Board(
    255.0,
    125.0,
    dpi=145,
    title="Vintage Scroll Templates: Portrait Lists & Landscape Roadmaps",
    background_style="dotgrid",
)
board.add_scroll_card(
    8.0,
    24.0,
    102.0,
    94.0,
    orientation="portrait",
    title="DFT Run Checklist",
    items=[
        ("1. Input", "Verify CIF / POSCAR"),
        ("2. K-mesh", "Check convergence"),
        ("3. Forces", "Relax < 0.01 eV/A"),
        ("4. SciComm", "Sketch with xkcd!"),
    ],
    bullet="checked",
    background_grid="ruled",
)
sc_land = board.add_scroll_card(
    116.0,
    24.0,
    132.0,
    94.0,
    orientation="landscape",
    title="Landscape Scroll: Synthesis Map",
    background_grid="dotgrid",
    inner_pad_mm=(4.0, 4.0, 14.0, 3.5),
)
xm.plot_pipeline_diagram(
    [("Mix", "Li2CO3 + TiO2"), ("Bake", "850 C"), ("XRD", "Cu Ka1")],
    ax=sc_land.axes,
    palette="pastel",
)
sc_land.axes.set_facecolor(xm.layout.scroll.PARCHMENT_LIGHT)
board.figure
```

### 4. Materials Informatics (Ternary Hull, Space Group Sunburst & Hatched Histogram)

Inspired by `pymatviz`, visualize ternary phase stability (`plot_ternary_phase_diagram`), International Tables space-group distributions (`plot_spacegroup_distribution`), and property histograms (`plot_histogram`) with zero external dependencies:

```python
import numpy as np

comps = [
    (1, 0, 0), (0, 1, 0), (0, 0, 1),
    (0.5, 0, 0.5), (0, 0.4, 0.6), (0.25, 0.25, 0.5),
    (0.5, 0.17, 0.33), (0.36, 0.36, 0.28),
]
energies = [0.0, 0.0, 0.0, -1.95, -1.65, -1.88, -2.02, -1.42]
labels = ["", "", "", "Li2O", "Fe2O3", "LiFeO2", "Li3FeO3", "meta"]
sgs = {225: 32, 227: 18, 194: 26, 139: 19, 62: 28, 14: 22, 2: 9}
rng = np.random.default_rng(42)
gaps = np.concatenate([rng.normal(1.3, 0.4, 120), rng.normal(3.1, 0.5, 70)])
gaps = gaps[gaps > 0.05]

with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(1, 3, figsize=(15.4, 4.8))
    xm.plot_ternary_phase_diagram(
        comps, energies, corners=("Li", "Fe", "O"), labels=labels,
        ax=axes[0], title="Li-Fe-O Ternary Hull",
    )
    xm.plot_spacegroup_distribution(
        sgs, mode="sunburst", shade="cycle",
        ax=axes[1], title="Space Group Sunburst",
    )
    xm.plot_histogram(
        gaps, bins=15, xlabel="Band Gap (eV)", shade="hatch",
        background_grid="dotgrid", ax=axes[2], title="Band-Gap Distribution",
    )
    fig.tight_layout()
```

### 5. xkcd MatChem Icons, Element Badges & Intuitive Crystal/Synthesis Graphs

Browse the built-in chemistry/materials icon set (`xm.plot_icon_sheet()`), drop individual icons anywhere with `xm.draw_icon(ax, "crystal", (x, y))`, compare a real 3D crystal side by side with its exact camera-matched GNN graph (`xm.plot_structure_graph(..., layout="camera", show_cell=True)`), and sketch synthesis/phase-transformation networks with rounded pill nodes (`node_shape="pill"`):

```python
with xm.xkcd_style("classic"):
    fig, axes = plt.subplots(1, 3, figsize=(15.4, 4.8), gridspec_kw={"width_ratios": [1.3, 1.0, 1.0]})
    xm.plot_icon_sheet(
        elements_to_show=("H", "Li", "C", "O", "Si", "Ti", "Fe", "Au"),
        ncols=6,
        ax=axes[0],
        title="xkcd MatChem Icons & Badges",
    )
    cam = xm.Camera.from_preset("isometric")
    xm.plot_crystal("si_diamond.cif", camera=cam, ax=axes[1], title="Actual Crystal (Si Diamond)")
    xm.plot_structure_graph(
        "si_diamond.cif",
        layout="camera",
        camera=cam,
        show_cell=True,
        ax=axes[2],
        title="Matching GNN Graph (Same View)",
    )
    fig.tight_layout()
```

```python
# Materials synthesis & phase-transformation network with rounded pill nodes
ax = xm.plot_graph_network(
    ["Anatase\nTiO2", "Rutile\nTiO2", "Li2CO3", "Li4Ti5O12\n(Spinel)"],
    [(0, 1, "650 C"), (0, 3, "800 C"), (2, 3, "+ Li")],
    pos=[(-1.2, 0.0), (0.1, 0.9), (0.0, -0.9), (1.3, 0.0)],
    node_shape="pill",
    directed=True,
    palette="pastel",
    title="Materials Synthesis Network (node_shape='pill')",
)
```
